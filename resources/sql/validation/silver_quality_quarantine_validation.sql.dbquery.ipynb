```sql
-- Validate that employee, assignment, and usage Silver valid outputs are clean,
-- and quarantine outputs preserve rejected records with standardized reasons.

-- ============================================================
-- Employee
-- ============================================================

SELECT
  'employee' AS dataset,
  quality_reason_text,
  COUNT(*) AS row_count
FROM adb_classic_compute_catalog.silver.employee_snapshot_quarantine_dlt
GROUP BY quality_reason_text
ORDER BY row_count DESC;

SELECT
  'employee_valid_has_quarantined_rows' AS check_name,
  COUNT(*) AS violation_count
FROM adb_classic_compute_catalog.silver.employee_snapshot_valid_dlt
WHERE is_quarantined = true;

SELECT
  'employee_quarantine_has_non_quarantined_rows' AS check_name,
  COUNT(*) AS violation_count
FROM adb_classic_compute_catalog.silver.employee_snapshot_quarantine_dlt
WHERE is_quarantined = false;

SELECT
  'employee_valid_missing_employee_id' AS check_name,
  COUNT(*) AS violation_count
FROM adb_classic_compute_catalog.silver.employee_snapshot_valid_dlt
WHERE employee_id IS NULL OR trim(employee_id) = '';

-- ============================================================
-- Assignment
-- ============================================================

SELECT
  'assignment' AS dataset,
  quality_reason_text,
  COUNT(*) AS row_count
FROM adb_classic_compute_catalog.silver.assignment_event_quarantine_dlt
GROUP BY quality_reason_text
ORDER BY row_count DESC;

SELECT
  'assignment_valid_has_quarantined_rows' AS check_name,
  COUNT(*) AS violation_count
FROM adb_classic_compute_catalog.silver.assignment_event_valid_dlt
WHERE is_quarantined = true;

SELECT
  'assignment_quarantine_has_non_quarantined_rows' AS check_name,
  COUNT(*) AS violation_count
FROM adb_classic_compute_catalog.silver.assignment_event_quarantine_dlt
WHERE is_quarantined = false;

SELECT
  'assignment_valid_missing_required_keys' AS check_name,
  COUNT(*) AS violation_count
FROM adb_classic_compute_catalog.silver.assignment_event_valid_dlt
WHERE assignment_event_id IS NULL
   OR trim(assignment_event_id) = ''
   OR assignment_id IS NULL
   OR trim(assignment_id) = '';

-- ============================================================
-- Usage
-- ============================================================

SELECT
  'usage' AS dataset,
  quality_reason_text,
  COUNT(*) AS row_count
FROM adb_classic_compute_catalog.silver.application_usage_event_quarantine_dlt
GROUP BY quality_reason_text
ORDER BY row_count DESC;

SELECT
  'usage_valid_has_quarantined_rows' AS check_name,
  COUNT(*) AS violation_count
FROM adb_classic_compute_catalog.silver.application_usage_event_valid_dlt
WHERE is_quarantined = true;

SELECT
  'usage_quarantine_has_non_quarantined_rows' AS check_name,
  COUNT(*) AS violation_count
FROM adb_classic_compute_catalog.silver.application_usage_event_quarantine_dlt
WHERE is_quarantined = false;

SELECT
  'usage_valid_missing_usage_event_id' AS check_name,
  COUNT(*) AS violation_count
FROM adb_classic_compute_catalog.silver.application_usage_event_valid_dlt
WHERE usage_event_id IS NULL OR trim(usage_event_id) = '';

-- ============================================================
-- reconciliation preview
-- ============================================================

SELECT 'employee' AS dataset, 'bronze' AS layer, COUNT(*) AS row_count
FROM adb_classic_compute_catalog.bronze.employee_directory_snapshot_raw_dlt
UNION ALL
SELECT 'employee' AS dataset, 'valid' AS layer, COUNT(*) AS row_count
FROM adb_classic_compute_catalog.silver.employee_snapshot_valid_dlt
UNION ALL
SELECT 'employee' AS dataset, 'quarantine' AS layer, COUNT(*) AS row_count
FROM adb_classic_compute_catalog.silver.employee_snapshot_quarantine_dlt
UNION ALL
SELECT 'assignment' AS dataset, 'bronze' AS layer, COUNT(*) AS row_count
FROM adb_classic_compute_catalog.bronze.license_assignment_change_event_raw_dlt
UNION ALL
SELECT 'assignment' AS dataset, 'valid' AS layer, COUNT(*) AS row_count
FROM adb_classic_compute_catalog.silver.assignment_event_valid_dlt
UNION ALL
SELECT 'assignment' AS dataset, 'quarantine' AS layer, COUNT(*) AS row_count
FROM adb_classic_compute_catalog.silver.assignment_event_quarantine_dlt
UNION ALL
SELECT 'usage' AS dataset, 'bronze' AS layer, COUNT(*) AS row_count
FROM adb_classic_compute_catalog.bronze.application_usage_event_raw_dlt
UNION ALL
SELECT 'usage' AS dataset, 'valid' AS layer, COUNT(*) AS row_count
FROM adb_classic_compute_catalog.silver.application_usage_event_valid_dlt
UNION ALL
SELECT 'usage' AS dataset, 'quarantine' AS layer, COUNT(*) AS row_count
FROM adb_classic_compute_catalog.silver.application_usage_event_quarantine_dlt;
```
